# Bank Customer Churn — Exploratory Data Analysis
**Google Colab-ready notebook**

This notebook loads the cleaned bank customer dataset and creates the main EDA visualizations used in the project.

```python
# STEP 1 — Upload the cleaned CSV to Google Colab
from google.colab import files
uploaded = files.upload()
```

```python
# STEP 2 — Load the dataset
import pandas as pd
import matplotlib.pyplot as plt

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# STEP 3 — Basic dataset information
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
# STEP 4 — Churn distribution
df["Exited"].value_counts().sort_index().plot(kind="bar", figsize=(6,4))
plt.title("Customer Churn Distribution")
plt.xlabel("Exited (0 = No, 1 = Yes)")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# STEP 5 — Churn rate by Geography
geo = df.groupby("Geography")["Exited"].mean().sort_values(ascending=False)
geo.plot(kind="bar", figsize=(6,4))
plt.title("Churn Rate by Geography")
plt.xlabel("Geography")
plt.ylabel("Churn Rate")
plt.show()
```

```python
# STEP 6 — Churn rate by Age
age = df.groupby("Age")["Exited"].mean()
age.plot(figsize=(7,4))
plt.title("Churn Rate by Age")
plt.xlabel("Age")
plt.ylabel("Churn Rate")
plt.show()
```

```python
# STEP 7 — Churn rate by Number of Products
products = df.groupby("NumOfProducts")["Exited"].mean()
products.plot(kind="bar", figsize=(6,4))
plt.title("Churn Rate by Number of Products")
plt.xlabel("Number of Products")
plt.ylabel("Churn Rate")
plt.show()
```

```python
# STEP 8 — Active vs Inactive Members
activity = df.groupby("IsActiveMember")["Exited"].mean()
activity.plot(kind="bar", figsize=(6,4))
plt.title("Churn Rate: Active vs Inactive Members")
plt.xlabel("Active Member (0 = No, 1 = Yes)")
plt.ylabel("Churn Rate")
plt.show()
```

```python
# STEP 9 — Gender analysis
gender = df.groupby("Gender")["Exited"].mean()
gender.plot(kind="bar", figsize=(6,4))
plt.title("Churn Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Churn Rate")
plt.show()
```

```python
# STEP 10 — Credit Score distribution by churn status
df.boxplot(column="CreditScore", by="Exited", figsize=(6,4))
plt.title("Credit Score by Churn Status")
plt.suptitle("")
plt.xlabel("Exited")
plt.ylabel("Credit Score")
plt.show()
```

```python
# STEP 11 — Tenure analysis
tenure = df.groupby("Tenure")["Exited"].mean()
tenure.plot(kind="bar", figsize=(7,4))
plt.title("Churn Rate by Tenure")
plt.xlabel("Tenure")
plt.ylabel("Churn Rate")
plt.show()
```

```python
# STEP 12 — Overall KPI summary
total = len(df)
churned = df["Exited"].sum()
churn_rate = df["Exited"].mean() * 100

print("Total Customers :", total)
print("Churned Customers:", churned)
print("Churn Rate       :", round(churn_rate, 2), "%")
print("Average Balance  :", round(df["Balance"].mean(), 2))
print("Average Credit Score:", round(df["CreditScore"].mean(), 2))
```
